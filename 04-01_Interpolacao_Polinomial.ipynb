
**🔹 Funções auxiliares**

As funções abaixo serão utilizadas ao longo da aula para construir polinômios interpoladores e gerar gráficos.


```python
def imprimir_matriz(M, nome="Matriz"):
    """
    Imprime uma matriz formatada.
    """

    print(f"{nome}:\n")

    for linha in M:

        for valor in linha:

            print(f"{valor:10.4f}", end=" ")

        print()

    print()


def imprimir_coeficientes(A):
    """
    Imprime os coeficientes do polinômio.
    """

    print("Coeficientes do polinômio:\n")

    for i, coef in enumerate(A):

        print(f"a{i} = {coef:.6f}")

    print()


def imprimir_polinomio(A):
    """
    Imprime o polinômio interpolador.

    Entrada:
    --------
    A = [a0, a1, a2, ..., an]
    """

    # Grau do polinômio
    n = len(A) - 1

    print(f"p_{n}(x) = ", end="")

    termos = []

    for i, coef in enumerate(A):

        coef = np.round(coef, 6)

        # Ignora coeficientes nulos
        if abs(coef) < 1e-10:
            continue

        # Termo constante
        if i == 0:

            termo = f"{coef:.6f}"

        # Termo linear
        elif i == 1:

            termo = f"{coef:.6f}x"

        # Demais termos
        else:

            termo = f"{coef:.6f}x^{i}"

        termos.append(termo)

    print(" + ".join(termos).replace("+ -", "- "))

    print()


def mostrar_resultados(X, Y, A):
    """
    Exibe:
    - pontos tabelados;
    - coeficientes;
    - polinômio interpolador.
    """

    print("Pontos tabelados:\n")

    for x, y in zip(X, Y):

        print(f"({x}, {y})")

    print()

    imprimir_coeficientes(A)

    imprimir_polinomio(A)


def plot_interpolacao(X, Y, P, titulo="Interpolação Polinomial"):
    """
    Gera o gráfico da interpolação.
    """

    x_plot = np.linspace(min(X)-1, max(X)+1, 400)

    y_plot = P(x_plot)

    plt.figure(figsize=(8,5))

    plt.plot(
        x_plot,
        y_plot,
        linewidth=2,
        label='Função interpoladora'
    )

    plt.scatter(
        X,
        Y,
        s=80,
        zorder=3,
        label='Pontos tabelados'
    )

    plt.xlabel('x')
    plt.ylabel('y')

    plt.title(titulo)

    plt.grid(True)
    plt.legend()

    plt.show()
    
def criar_polinomio(A):
    """
    Cria o polinômio interpolador a partir
    do vetor de coeficientes.

    Entrada:
    --------
    A = [a0, a1, a2, ..., an]

    Retorna:
    --------
    P : polinômio interpolador
    """

    return np.poly1d(np.flip(A))
```


## $ \S 1 $ Introdução

A interpolação polinomial é um processo em que aproximamos uma função $f$ por um polinômio $P$, chamado **polinômio interpolador**.

📌 Essa técnica é muito utilizada quando:

- quando são conhecidos somente os valores numéricos da função para um conjunto de pontos e é necessário calcular o valor da função em outros pontos;
- quando a função em estudo tem uma expressão tal que operações como integração ou diferenciação são difíceis (ou impossíveis) de realizar diretamente.


## $ \S 2 $ Interpolação polinomial

Dados os pontos

$$
(x_0,f(x_0)),\ (x_1,f(x_1)),\ \ldots,\ (x_n,f(x_n)),
$$

portanto $(n+1)$ pontos, queremos aproximar $f(x)$ por um polinômio
$p_n(x)$, de grau menor ou igual a $n$, tal que

$$
f(x_k)=p_n(x_k),
\qquad k=0,1,2,\ldots,n.
$$

Surgem então as perguntas:

- Existe sempre um polinômio $p_n(x)$ que satisfaça essas condições?
- Caso exista, ele é único?

Representaremos $p_n(x)$ por:

$$
p_n(x)=a_0+a_1x+a_2x^2+\cdots+a_nx^n.
$$

Portanto, obter $p_n(x)$ significa determinar os coeficientes

$$
a_0,a_1,\ldots,a_n.
$$

Da condição

$$
p_n(x_k)=f(x_k),
\qquad \forall k=0,1,2,\ldots,n,
$$

obtemos o seguinte sistema linear:

$$
\begin{cases}
a_0+a_1x_0+a_2x_0^2+\cdots+a_nx_0^n=f(x_0) \\\\
a_0+a_1x_1+a_2x_1^2+\cdots+a_nx_1^n=f(x_1) \\\\
\vdots \\\\
a_0+a_1x_n+a_2x_n^2+\cdots+a_nx_n^n=f(x_n)
\end{cases}
$$

com $(n+1)$ equações e $(n+1)$ incógnitas.

$$
\underbrace{
\begin{bmatrix}
1 & x_0 & x_0^2 & \cdots & x_0^n\\
1 & x_1 & x_1^2 & \cdots & x_1^n\\
\vdots & \vdots & \vdots & \ddots & \vdots\\
1 & x_n & x_n^2 & \cdots & x_n^n
\end{bmatrix}
}_{V}
\begin{bmatrix}
a_0\\a_1\\ \vdots\\a_n
\end{bmatrix}
=
\begin{bmatrix}
y_0\\y_1\\ \vdots\\y_n
\end{bmatrix}.
$$

A matriz $V$ é chamada de **matriz de Vandermonde**.

Desde que $x_0,x_1,\ldots,x_n$ sejam pontos distintos, temos que $\det(V)\neq 0,$ e, então, o sistema linear admite solução única.

<div style="background-color: #bdffb0; padding: 12px; border-radius: 6px;">

**📐Teorema 4-01.1** Existe um único polinômio $p_n(x),$ de grau $\leq n,$ tal que: $p_n(x_k) = f(x_k)$, para $k = 0, 1, \ldots, n$ desde que $x_k\neq x_j$ para $k\neq j$.

</div>

## $ \S 3 $ Formas de se obter $p_n(x)$

Existem diversas formas de se obter o polinômio interpolador $p_n(x)$. Veremos três formas:

### $ \S 3.1 $ Resolução do sistema linear 


```python
def vandermonde(X, Y):
    """
    Calcula os coeficientes do polinômio interpolador utilizando a matriz de Vandermonde.

    Parâmetros
    ----------
    X : vetor com os valores de x
    Y : vetor com os valores de y

    Retorna
    -------
    A : coeficientes do polinômio interpolador
    V : matriz de Vandermonde
    """

    # Número de pontos
    n = len(X)

    # Inicializa a matriz de Vandermonde
    V = np.zeros((n, n))

    # Preenche a matriz
    for i in range(n):

        for j in range(n):

            V[i, j] = X[i]**j

    # Resolve o sistema linear
    A = np.linalg.solve(V, Y)

    return A, V
```

### $ \S 3.2 $ Forma de Lagrange

Sejam $x_0,x_1,\ldots,x_n$ pontos distintos e $y_k=f(x_k)$, para $k=0,1,\ldots,n$.

A forma de Lagrange para o polinômio interpolador é dada por

$$
p_n(x)=\sum_{k=0}^{n} y_kL_k(x),
$$

onde

$$
L_k(x)=
\frac{
\displaystyle \prod_{\substack{j=0 \\\\ j\neq k}}^{n}(x-x_j)
}{
\displaystyle \prod_{\substack{j=0 \\\\ j\neq k}}^{n}(x_k-x_j)
}.
$$

Os polinômios $L_k(x)$ são chamados de **polinômios básicos de Lagrange (Fatores de Lagrange)**.



### 🔹 Implementação em Python

```python
def lagrange(X, Y, x):
    """
    Calcula o valor do polinômio interpolador
    de Lagrange no ponto x.

    Parâmetros
    ----------
    X : vetor com os valores de x
    Y : vetor com os valores de y
    x : ponto onde deseja-se interpolar

    Retorna
    -------
    y : valor interpolado em x
    """

    # Inicializa o valor do polinômio
    y = 0

    # Número de pontos
    n = len(X)

    # Soma dos termos de Lagrange
    for k in range(n):

        # Inicializa Lk(x)
        Lk = 1

        # Produto para i < k
        for i in range(k):

            xi = X[i]
            xk = X[k]

            Lk = Lk * (x - xi) / (xk - xi)

        # Produto para i > k
        for i in range(k + 1, n):

            xi = X[i]
            xk = X[k]

            Lk = Lk * (x - xi) / (xk - xi)

        # Valor correspondente de y
        yk = Y[k]

        # Soma Yk * Lk(x)
        y = y + Lk * yk

    return y
```

### $ \S 3.3 $ Método de Newton

O método de Newton utiliza diferenças divididas para construir o polinômio interpolador.

A forma geral é

$$
p_n(x)=c_0+(x-x_0)c_1+(x-x_0)(x-x_1)c_2+\cdots+(x-x_0)(x-x_1)\cdots(x-x_{n-1})c_n.
$$

📌 Vantagens:
- Mais simples de calcular "na mão" do que o sistema linear de Vandermonde.
- É incremental, ou seja, fácil atualização quando novos pontos são adicionados.

Nosso trabalho é encontrar os coeficientes

$$
c_0,c_1,\ldots,c_n.
$$

Esses coeficientes são chamados de **diferenças divididas**.


Usamos a notação

$$
f[x_i,\ldots,x_j]
$$

para representar as **diferenças divididas**.

Os coeficientes do polinômio de Newton são dados por

$$
c_k=f[x_0,x_1,\ldots,x_k].
$$


🟩 Ordem 0

$$
c_0=f[x_0]=y_0
$$

🟩 Ordem 1

$$
c_1=f[x_0,x_1]
=
\frac{f[x_1]-f[x_0]}{x_1-x_0}
=
\frac{y_1-y_0}{x_1-x_0}
$$

🟩 Ordem 2

$$
c_2=f[x_0,x_1,x_2]
=
\frac{
f[x_1,x_2]-f[x_0,x_1]
}{
x_2-x_0
}
$$

🟩 Ordem $k$ (forma geral)

$$
c_k=f[x_0,\ldots,x_k]
=
\frac{
f[x_1,\ldots,x_k]
-
f[x_0,\ldots,x_{k-1}]
}{
x_k-x_0
}
$$

📌 Isso pode parecer complicado inicialmente, mas o cálculo torna-se bastante simples utilizando uma tabela de diferenças divididas.

</div>

🧪[__Exemplo 4-03.2__](#exemplo-4-03.2) Usando a forma de Newton, o polinômio $p_2(x),$ que interpola $f(x)$ nos pontos dados abaixo:

| $x$ | $-1$ | $0$ | $2$ |
| :--- | :--- | :--- | :--- |
| $f(x)$ | $4$ | $1$ | $-1$ |

é dado por

$$p_2(x) = f[x_0] +(x-x_0)f[x_0,x_1] + (x-x_0)(x-x_1)f[x_0,x_1,x_2].$$

_Solução:_

🔹 Tabela de diferenças divididas

$$
\begin{array}{c|c c c}
x & \text{Ordem 0} & \text{Ordem 1} & \text{Ordem 2} \\
\hline
-1 & 4 & & \\
\\
& & -3 & \\
\\
0 & 1 & & \dfrac{2}{3} \\
\\
& & -1 & \\
\\
2 & -1 & &
\end{array}
$$

Substituindo os coeficientes na fórmula de Newton:

$p_2(x) = 4+ (x+1)(-3) + (x+1)(x-0)\frac{2}{3}.$

Observamos que, agrupando os termos semelhantes, obtemos $p_2(x) = \frac{2}{3}x^2 - \frac{7}{3}x + 1.$

```python
def newton_interp(X, Y, px=None):
    """
    Interpolação polinomial pelo método de Newton.

    Parâmetros
    ----------
    X : vetor com os valores de x
    Y : vetor com os valores de y
    px : ponto onde deseja-se avaliar o polinômio (opcional)

    Retorna
    -------
    C : coeficientes de Newton
    y_eval : valor interpolado em px (opcional)
    """

    # Converte para arrays numpy
    X = np.array(X, dtype=float)

    Y = np.array(Y, dtype=float)

    # Número de pontos
    n = len(X)

    # ---------------------------------
    # Diferenças divididas
    # ---------------------------------

    # Inicializa os coeficientes
    C = Y.copy()

    # Construção da tabela
    for j in range(1, n):

        for i in range(n - 1, j - 1, -1):

            numerador = C[i] - C[i - 1]

            denominador = X[i] - X[i - j]

            C[i] = numerador / denominador

    # ---------------------------------
    # Exibição dos resultados
    # ---------------------------------

    print("--------------------------------------------------")
    print("Forma de Newton para o polinômio interpolador")
    print("--------------------------------------------------\n")

    # Coeficientes
    print("Coeficientes de Newton:\n")

    for i in range(n):

        print(f"c{i} = {C[i]:.6f}")

    print()

    # ---------------------------------
    # Impressão do polinômio
    # ---------------------------------

    print("Polinômio na forma de Newton:\n")

    print(f"P(x) = {C[0]:.6f}", end="")

    for i in range(1, n):

        print(f" + ({C[i]:.6f})", end="")

        for j in range(i):

            print(f"(x - ({X[j]}))", end="")

    print("\n")

    # ---------------------------------
    # Avaliação do polinômio
    # ---------------------------------

    if px is not None:

        # Inicializa com o último coeficiente
        y_eval = C[n - 1]
        
        for k in range(n - 2, -1, -1):

            y_eval = y_eval * (px - X[k]) + C[k]

        print("Avaliação do polinômio:\n")

        print(f"P({px}) = {y_eval:.6f}\n")

        return C, y_eval

    return C
```

```python
# Pontos tabelados
X = [-1, 0, 2]

Y = [4, 1, -1]

# Avaliação em x = 1
C, y = newton_interp(X, Y, px=1)
```
