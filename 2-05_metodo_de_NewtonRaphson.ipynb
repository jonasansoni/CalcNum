## $ \S 1 $ Introdução

No estudo do <span style="color:green;">__Método do Ponto Fixo (MPF)__</span>, vimos que:

 i) uma das condições de convergência é que $|\varphi'(x)| \le M < 1, \quad \forall x \in I,$ onde $I$ é um intervalo centrado na raiz;

 ii) a convergência do método será mais rápida quanto menor for $|\varphi'(\xi)|$.

<div style='background-color:#f0f0f0; padding:20px; border-radius:8px'>

🎯 O que o **método de Newton** faz, na tentativa de garantir e acelerar a convergência do MPF, é escolher para função de iteração uma função $\varphi(x)$ tal que $\varphi'(\xi) = 0$.
</div>

Então, dada a equação $f(x)=0$ e partindo da forma geral para $\varphi(x)$, queremos obter uma função $A(x)$ tal que $\varphi'(\xi)=0$.

$$
\varphi(x) = x + A(x)f(x)
$$

$$
\Rightarrow \varphi'(x) = 1 + A'(x)f(x) + A(x)f'(x)
$$

$$
\Rightarrow \varphi'(\xi) = 1 + A'(\xi)f(\xi) + A(\xi)f'(\xi)
$$

Como $f(\xi)=0$, segue que:
$$
\varphi'(\xi) = 1 + A(\xi)f'(\xi).
$$

Assim,
$$
\varphi'(\xi)=0 \iff 1 + A(\xi)f'(\xi)=0 \;\Rightarrow\; A(\xi)= -\frac{1}{f'(\xi)}.
$$

Tomamos então:
$$
A(x)= -\frac{1}{f'(x)}.
$$

Logo, a função de iteração é:
$$
\color{blue}\varphi(x) = x - \frac{f(x)}{f'(x)}.
$$

Derivando:
$$
\varphi'(x) = 1 - \frac{[f'(x)]^2 - f(x)f''(x)}{[f'(x)]^2}
= \frac{f(x)f''(x)}{[f'(x)]^2}.
$$

Como $f(\xi)=0$, temos:
$$
\varphi'(\xi)=0 \quad (\text{desde que } f'(\xi)\ne 0).
$$


Assim, escolhido $x_0$, a sequência $\{x_k\}$ é definida por:
$$\color{blue} x_{k+1} = x_k - \frac{f(x_k)}{f'(x_k)}, \quad k=0,1,2,\dots$$


## $ \S 2 $ Motivação geométrica

Suponha que desejemos encontrar um zero de uma função diferenciável $f$. Partindo de uma estimativa inicial $ x_0 $, a idéia por trás do __método de Newton__ é substituir a função pela sua reta tangente na estimativa atual $x_k$ e tomar a intersecção desta reta com o eixo-$x$ como valor da próxima estimativa $ x_{k + 1} $.  Em símbolos, a reta tangente ao gráfico de $ f $ em $x_k$ é descrita pela equação:
$$
    y = f(x_k) + f'(x_k)\,(x - x_{k})\,. \tag{1}
$$
A aproximação $ x_{k + 1} $ seguinte é obtida fazendo $ y = 0 $ e encontrando o valor de $ x $ correspondente:
$$
\color{blue}{
\boxed{x_{k + 1} = x_k - \frac{f(x_k)}{f'(x_k)}}
} \tag{2}
$$

📝 A expressão do lado direito em (1) é o polinômio de Taylor de grau $ 1 $ de $ f $ ao redor de $ x_k $, às vezes chamado nos cursos de Cálculo de _linearização_ ou _aproximação linear_ de $f$ aí. Portanto, sob outro ponto de vista, completamente equivalente ao que descrevemos antes, o método de Newton consiste em substituir a função original $ f $ pelo seu polinômio de Taylor de grau $1$ ao redor da estimativa atual e usar o zero deste como nova aproximação para o zero de $f$.

📝 O método de Newton também é conhecido como _método de Newton-Raphson_; [Joseph Raphson (1668–1715)](https://en.wikipedia.org/wiki/Joseph_Raphson) foi um matemático inglês contemporâneo de Newton.

🧪[__Exemplo 2-05.1__](#exemplo-2-05.1) Rode em seqüência o código nas três células abaixo para ver uma animação do método de Newton aproximando um zero da função $$ f(x) =  x^3 - 2x - 3 \ln x .$$ (Para facilitar a navegação posterior, depois de executar estas células, você pode escondê-las.) 

````python
def print_solution(xs: list[float], ys: list[float], freq: int = 1) -> None:
    """
    Given two arrays xs and ys of the same length, prints a table whose k-th
    line consists of three entries: the values of k, xs[k] and ys[k].
    Parameters:
        * The arrays xs and ys. 
        * A parameter freq used to print only one in every freq line. The
          first and last line are always printed. If freq == 0, then only these
          lines are printed.
    Returns: None.
    Prints: A header and the table described above.
    """
    def print_header() -> None:
        """
        Prints the table's header.
        """
        print("\n|       k      ", end="")
        print("    x_k            ", end="")
        print("    f(x_k)      |")
        print("|=================================================|")
        
    def print_line(k: int, x: float, y: float) -> None:
        """
        Pretty-prints k, x and y.
        """
        print(f"|      {k:02}", end="")
        print(f"    {x:15.8f}", end="")
        print(f"    {y:15.8f}   |")
    
    
    N = len(xs)
    if freq == 0:       # If freq == 0, print only first and last lines.
        freq = N - 1
    print_header()
    print_line(0, xs[0], ys[0])
    for n in range(1, N, freq):
        print_line(n, xs[n], ys[n])
    if n != N - 1:
        print_line(n, xs[N], ys[N])
    print("|_________________________________________________|\n")
        
    return None
````

````python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
from typing import Callable, List, Tuple

def newton_animation(f: Callable[[float], float],
                     df: Callable[[float], float],
                     a: float, b: float, x: float,
                     N: int = 4, title: str = "", duration: float = 0.75
                     ) -> Tuple[List[float], List[float], HTML]:
    """
    Displays an animation of Newton's method applied to a function.

    Parameters:
    * f: A differentiable real function f.
    * df: Its derivative df (as a function).
    * a: The left endpoint of the interval in the x-axis where the animation takes place.
    * b: The right endpoint of the interval in the x-axis where the animation takes place.
    * x: An initial estimate x for the zero in [a, b].
    * N: The maximum number of iterations.
    * title: A title to be displayed at the top of the diagram.
    * duration: The duration of the pause between slides, in seconds.

    Returns:
    * xs: List of float, the estimates from Newton's method.
    * ys: List of float, the values of the function f at each of the estimates.
    * animation_html: IPython.display.HTML object containing the animation.

    Displays:
    * The animation in a Jupyter notebook cell.
    """

    def tangent_line_at(x_0):
        return lambda x: f(x_0) + df(x_0) * (x - x_0)
    
    def iteration(x):
        return x - f(x) / df(x)

    cmap = plt.get_cmap("tab10")  # Used to control the colors.
    P = 200  # Number of points in each plot.
    width = 1.75  # Line width.
    marker_size = 5
    domain = np.linspace(a, b, P)  # Generates P nodes from a to b.
    xs = [x]  # Stores the estimates.
    for _ in range(N):  # Filling xs.
        xs.append(iteration(xs[-1]))
    ys = [f(x) for x in xs]  # Stores f of the estimates.
    tangent_lines = [tangent_line_at(x) for x in xs]

    # Generate sample points for x-intervals between consecutive estimates:
    xs_range = [np.linspace(xs[n], xs[n + 1], P) for n in range(N)]
    # Lists containing the x and y coordinates for plotting vertical lines:
    xs_vert = [np.linspace(xs[n], xs[n], P) for n in range(N)]
    ys_vert = [np.linspace(0, ys[n], P) for n in range(N)]

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.axhline(y=0.0, color='black', linestyle='-', lw=width)
    ax.plot(domain, f(domain), label="$ y = f(x) $", lw=width)
    ax.set_xlabel("$ x $")
    ax.set_ylabel("$ y $")
    ax.set_title(title)
    ax.grid(linestyle="dotted", alpha=1.0)

    lines = []
    for i in range(N + 1):
        color = cmap(i + 1)
        if i < N:
            tangent_color = cmap(i + 2)
        else:
            tangent_color = color
        label = f"$ x_{i} $" if i == 0 else None
        (line,) = ax.plot([], [], color=color, marker="x", mew=width, label=label)
        (vert_line,) = ax.plot([], [], linestyle='dotted', lw=width, color='black')
        (point,) = ax.plot([], [], color='black', marker="o", ms=marker_size)
        (tangent_line,) = ax.plot([], [], linestyle='--', lw=width, color=tangent_color)
        lines.append((line, vert_line, point, tangent_line))

    def init():
        for line, vert_line, point, tangent_line in lines:
            line.set_data([], [])
            vert_line.set_data([], [])
            point.set_data([], [])
            tangent_line.set_data([], [])
        return [item for sublist in lines for item in sublist]

    def update(frame):
        if frame == 0:
            lines[0][0].set_data([xs[0]], [0])
            return [item for sublist in lines for item in sublist]
        n = (frame - 1) // 3
        substep = (frame - 1) % 3
        if n < N:
            x = xs[n]
            y = ys[n]
            x_next = xs[n + 1]
            line, vert_line, point, tangent_line = lines[n]
            next_line = lines[n + 1][0]
            if substep == 0:
                vert_line.set_data(xs_vert[n], ys_vert[n])
                point.set_data([x], [y])
            elif substep == 1:
                tangent_line.set_data(xs_range[n], tangent_lines[n](xs_range[n]))
            elif substep == 2:
                next_line.set_data([x_next], [0])
                next_line.set_label(f'$ x_{n + 1} $')
        elif frame == N * 3 + 1:
            ax.legend()
        elif frame == N * 3 + 2:
            return [item for sublist in lines for item in sublist]
        return [item for sublist in lines for item in sublist]

    ani = FuncAnimation(fig, update, frames=range(N * 3 + 3), init_func=init,
                        blit=True, interval=duration * 250)
    
    plt.close(fig)
    return xs, ys, HTML(ani.to_jshtml())
````

````python
from numpy import log
f = lambda x: x**3 - 2 * x - 3 * log(x)
df = lambda x: 3 * x**2 - 2 - 3 / x
a = 0.4
b = 1.5
x_0 = 1
N = 3
pausa = 3.0
titulo = "Exemplo: Método de Newton para $ y = x^3 - 2x - 3\\,\\ln x $."\
         "\n Estimativa inicial $ x_0 = 1 $ e $ N = 3 $ iterações."

xs, ys, ani = newton_animation(f, df, a, b, x_0, N, titulo, pausa)
print_solution(xs, ys)
ani
````

## $ \S 2 $ Estudo da convergência do método de Newton
<div style="background-color:#bdffb0; padding:12px; border-radius:6px;">

📐 **Teorema 2-05.1**

Sejam $f(x)$, $f'(x)$ e $f''(x)$ contínuas em um intervalo $I$ que contém a raiz $x=\xi$ de $f(x)=0$. Suponha que
$$
\color{red}{f'(\xi) \ne 0}.
$$

Então, existe um intervalo $\bar{I} \subset I$, contendo a raiz $\xi$, tal que, se $x_0 \in \bar{I}$, a sequência $\{x_k\}$ gerada pela fórmula recursiva
$$
x_{k+1} = x_k - \frac{f(x_k)}{f'(x_k)}
$$
converge para a raiz.
</div>

<div style="background-color: #fff3b0; padding: 12px; border-radius: 6px;">

⚠️ **Observação:** A demonstração deste teorema pode ser consultada em [Ruggiero & Lopes, 1996](#ref-ruggiero).
</div>

<div style="background-color: #D1FFBD; padding: 12px; border-radius: 6px;">

✅ **Conclusão:** Em geral, afirma-se que o método de Newton converge desde que a estimativa inicial $x_0$ seja escolhida suficientemente próxima da raiz $\xi$.

</div>

Comprovaremos neste exemplo que uma escolha cuidadosa da aproximação inicial é, em geral, essencial para o bom desempenho do método de Newton.

## $\S 3$ Critérios de parada 

Ao aplicar um **método iterativo**, obtém-se uma sequência $\{x_k\}$ que, sob hipóteses adequadas, converge para uma raiz $\xi$ da equação $f(x)=0$. 

Como, em geral, a raiz exata $\xi$ é desconhecida, torna-se necessário estabelecer **critérios de parada**, isto é, condições baseadas em quantidades computáveis que permitam interromper o processo iterativo quando se atinge uma precisão previamente fixada.

Dentre os critérios mais utilizados, destacam-se:

- $\left| x_{k+1} - x_k \right| < \varepsilon_1 \quad$ (_erro absoluto_);

- $\left| f(x_{k+1}) \right| < \varepsilon_2 \quad$ (_resíduo_);

- $\dfrac{\left| x_{k+1} - x_k \right|}{\left| x_{k+1} \right|} < \varepsilon_3 \quad$ (_erro relativo_);

onde $\varepsilon_1, \varepsilon_2, \varepsilon_3 > 0$ são tolerâncias previamente estabelecidas.

> **Observação:** A condição $|x_{k+1} - x_k| < \varepsilon_1$ não implica, em geral, que $|x_{k+1} - \xi| < \varepsilon_1$, pois o erro verdadeiro depende da distância até a raiz exata, a qual é desconhecida. Assim, critérios de parada baseados apenas na diferença entre iterados fornecem apenas uma indicação indireta da precisão da aproximação.

> Na prática, é comum empregar mais de um critério de parada simultaneamente, de modo a aumentar a confiabilidade dos resultados obtidos.

## $ \S 4 $ Algoritmo do método de Newton

<div style="border:2px solid #444; padding:15px; border-radius:8px; background:#f7f7f7">

### Algoritmo — Método de Newton

**Entrada**

- Função $f(x)$
- Derivada $f'(x)$
- Aproximação inicial $x_0$
- Tolerâncias $\varepsilon_1$ e $\varepsilon_2$

**Saída**

- Aproximação $\bar{x}$ da raiz de $f(x)=0$

---

**Passos**

1. Dados iniciais:
   - $x_0$: aproximação inicial;
   - $\varepsilon_1$, $\varepsilon_2$: precisões.

2. Se $|f(x_0)| < \varepsilon_2$, faça $\bar{x} = x_0$. **FIM**

3. $k \leftarrow 1$

4. Calcule
$$
x_1 = x_0 - \frac{f(x_0)}{f'(x_0)}
$$
   
5. Se $|f(x_1)| < \varepsilon_2$ ou $|x_1 - x_0| < \varepsilon_1$,  
   faça $\bar{x} = x_1$. **FIM**

6. $x_0 \leftarrow x_1$

7. $k \leftarrow k + 1$  
   Volte ao passo **4**

---

Adaptado de: [Ruggiero & Lopes, 1996](#ref-ruggiero).

</div>

````python
def newton_eq(f, df, x0, eps1, eps2, K):
    """
    Método de Newton — Algoritmo

    Entrada
    -------
    f  : função f(x)
    df : derivada f'(x)
    x0 : aproximação inicial
    eps1 : tolerância para |f(x)|
    eps2 : tolerância para |x_k - x_{k-1}|
    K  : número máximo de iterações

    Saída
    -----
    x̄ : aproximação da raiz de f(x) = 0
    
    Observações
    -----------
    - Requer f'(x_k) ≠ 0.
    - A convergência é local.
    - Sob condições adequadas, a convergência é quadrática.
    """

    print(f"{'k':>2} {'x_k':>14} {'f(x_k)':>14} {'|x_k-x_{k-1}|':>14}")

    # Passo 1
    if abs(f(x0)) < eps2:
        print(f"{0:2d} {x0:14.6f} {f(x0):14.6e} {'-':>14}")
        return x0

    print(f"{0:2d} {x0:14.6f} {f(x0):14.6e} {'-':>14}")

    # Passo 2
    k = 1

    while True:

        # Segurança
        if df(x0) == 0:
            raise ValueError("Derivada nula — método de Newton falha.")

        # Passo 3
        x1 = x0 - f(x0) / df(x0)

        erro = abs(x1 - x0)

        print(f"{k:2d} {x1:14.6f} {f(x1):14.6e} {erro:14.6e}")

        # Passo 4
        if abs(f(x1)) < eps2 or erro < eps1:
            print("-" * 78)
            print(f"Total de iterações = {k}")
            return x1

        # Passo 5 (extra prático)
        if k >= K:
            print("Número máximo de iterações atingido.")
            return x1

        # Passo 6
        x0 = x1

        # Passo 7
        k += 1
````

## $ \S 4 $ Estudo da convergência do método de Newton


<div style="background-color: #ffffbd; padding: 12px; border-radius: 6px;">

💡 <b>Observação</b>: O método de Newton tem, tipicamente, convergência extremamente rápida sempre que o ponto inicial é uma boa aproximação da solução desejada.
</div>

Um primeiro exemplo disso, que vamos explorar, é pensar nesse método sendo usado para calcular a raiz quadrada de um número. O problema de calular a raiz de um número $a > 0$ dado pode ser visto como o problema de resolver a equação $f(x) = x^2 - a = 0.$
Nesse caso é muito fácil calcular a derivada. Assim, a iteração
$$
x_{k + 1} = x_k - \frac{f(x_k)}{f'(x_k)}
$$
pode ser escrita como 
$$
x_{k + 1} = x_k - \frac{x_k^2 - a}{2 x_k} = \frac{1}{2}\left(x_k + \frac{a}{x_k} \right).
$$
Vamos implementar esse algoritmo e tentar calcular $\sqrt{10}$ partindo de $x_0 = 1$.

````python
import math

a = 10 # Valor do qual desejamos calcular a raiz quadrada
xk = 1 # Aproximação inicial para √a
prec = 1.0e-14 # Tolerância (critério de parada baseado no erro relativo)

iters = 0 # Contador de iterações
print(iters, ":", xk) # Impressão da aproximação inicial

while abs((xk*xk - a)/a) > prec: # Iteração do método de Heron/Newton
    xk = 0.5*(xk + a/xk) # Fórmula iterativa do método
    iters += 1 # Atualiza o contador de iterações
    print(iters, ":", xk) # Mostra a nova aproximação obtida

print("\nValor 'exato' =", math.sqrt(a)) # Comparação com o valor calculado pela função padrão
print("xk*xk        =", xk*xk) # Verificação do quadrado da aproximação final
````

Note que a convergência ocorre de forma extremamente rápida. Na iteração 2, o número já foi calculado com 1 casa correta, na próxima iteração o número de casa corretas já duplicou passando para 2, depois para 4, depois para 8 casas e por fim 15. Ou seja, o número de casas corretas dobrou aproximadamente por iteração. Esse é um exemplo de convergência quadrática, que é a taxa de convergência típica do método de Newton.

<div style='background-color:#f0f0f0; padding:20px; border-radius:8px'>

🎯 Portanto, o __método de Newton tem convergência quadrática__.

 <b>Conclusão:</b> Uma vez suficientemente próxima da raiz, a sequência gerada pelo método de Newton converge quadraticamente, isto é, o erro em cada iteração é proporcional ao quadrado do erro anterior, o que explica o rápido aumento do número de casas decimais corretas.

📌 Em particular, essa convergência é local, dependendo da escolha de uma aproximação inicial suficientemente próxima da raiz.

## § 5 — Vantagens e desvantagens do método de Newton

<span style="color:blue;">**Vantagens**</span>
- 😊 Convergência quadrática (mais rápida que métodos como bissecção).

<span style="color:red;">**Desvantagens**</span>
- 😕 A escolha de uma boa aproximação inicial nem sempre é simples.
- 😕 Necessidade de calcular $f'(x)$ a cada iteração.
- 😕 Pode não convergir se a aproximação inicial estiver longe da raiz.
- 😕 Pode falhar se $f'(x_k)$ for muito pequeno ou nulo.

📝[__Exercício 2-05.6__](#exercicio-2-05.6) — __Método da Secante__

Uma grande desvantagem do método de Newton é a necessidade de se obter $ f'(x) $ e calcular seu valor numérico a cada iteração.

Uma forma de se contornar este problema é substituir a derivada $ f'(x_k) $ pelo quociente das diferenças:

$$
f'(x_k) \approx \frac{f(x_k) - f(x_{k-1})}{x_k - x_{k-1}}
$$

onde $ x_k $ e $ x_{k-1} $ são duas aproximações para a raiz.

Neste caso, a função de iteração fica

$$
\varphi(x_k) = x_k - \frac{f(x_k)}{\dfrac{f(x_k) - f(x_{k-1})}{x_k - x_{k-1}}}
$$

$$
= x_k - \frac{f(x_k)}{f(x_k) - f(x_{k-1})}(x_k - x_{k-1})
$$

Ou ainda,

$$
\varphi(x_k) = \frac{x_{k-1} f(x_k) - x_k f(x_{k-1})}{f(x_k) - f(x_{k-1})}
$$
