## $ \S 1 $ Existência e unicidade de soluções

Um resultado de cáculo fundamental para tratar da existência de soluções de uma equação não linear é o **Teorema de Bolzano**.

<div style="background-color: #bdffb0; padding: 12px; border-radius: 6px;">

**📐Teorema 2-02.1 (Teorema de Bolzano)** Seja $f: [a,b] \rightarrow \mathbb{R}$ uma função contínua, tal que **$f(a)f(b) < 0$.** Então existe $c \in (a, b)$, tal que $f(c) = 0$.

</div>

Ou seja, se uma função contínua troca de sinal nos extremos em um intervalo, então ela possui pelo menos um zero (nesse intervalo). 

Já para garantir a unicidade é preciso exigir mais da função $f$. Uma hipótese razoável é que ela seja constantemente crescente ou decrescente dentro do intervalo. Para isso, basta exigir que a derivada da função não troque de sinal.

<div style="background-color: #bdffb0; padding: 12px; border-radius: 6px;">

**📐Corolário 2-02.1** Seja $f: [a,b] \rightarrow \mathbb{R}$ uma função diferenciável em $(a, b)$. Se $f(a)f(b) < 0$ e $f'(x)$ não troca de sinal em $(a, b)$, então existe um único $c \in (a, b)$, tal que $f(c) = 0$.
</div>

Note que temos que considerar os valores da derivada em todo o intervalo, e não apenas nos extremos.

# $ \S 2 $ Método da bissecção

O Teorema de Bolzano serve de ponto de partida para um primeiro método iterativo para resolução de equações não-lineares conhecido como bissecção. A ideia dele é simples.

## $ \S 2.1 $ Descrição do método da bissecção

Como acima, sejam $ f $ uma função real _contínua_ e $ [a, b] $ um intervalo dentro do domínio de $ f $ tal que
$$
\operatorname{sinal} f(a) \ne \operatorname{sinal} f(b)
$$
No **método da bissecção** começamos tomando $m$ como o <span style="color:blue">**ponto médio**</span> de $ a $ e $ b $,
$$
m = \frac{a+b}{2} 
$$
e avaliando $ f $ aí. Há três possibilidades:
1. Se $ f(m)= 0 $, então $ m $ é um zero de $ f $ e podemos terminar.

2. Se $ \operatorname{sinal} f(m)  \neq \operatorname{sinal} f(a) $, então $ f $ troca de sinal em $ [a, m] $. Neste caso fazemos $ b \leftarrow m $.

3. Se $ \operatorname{sinal} f(m) = \operatorname{sinal} f(a) $, então $ f $ troca de sinal em $ [m, b] $. Neste caso fazemos $ a \leftarrow m $.

<div style="background-color: #fff3b0; padding: 12px; border-radius: 6px;">

⚠️ Nos dois últimos casos, podemos restringir nossa busca a um intervalo cujo comprimento é a **metade** do anterior. O teorema de Bolzano garante
que este subintervalo ainda contém um zero, portanto podemos aplicar o mesmo procedimento a ele, usando os valores atualizados de $ a $ e $ b $.  Bissecções sucessivas eventualmente nos levarão a um zero ou a um encaixotamento de um zero por um subintervalo de comprimento tão pequeno quanto desejado.  Quando isto acontecer, retornamos o ponto médio deste subintervalo como estimativa para o zero.
</div>

✅ O método da bissecção também é conhecido como _busca binária_.

✅ Não é necessário _isolar_ um zero dentro do intervalo inicial, ou seja, o método funciona mesmo que haja mais de um zero dentro de $ [a, b] $.

🧪[__Exemplo 2-02.1__](#exemplo-2-02.1) Execute em sequência as três células abaixo para ver uma animação do método da bissecção aproximando um zero da função $ f(x) = x^3 - x - 3 $. (**Para facilitar a navegação posterior, depois você pode esconder as células clicando sobre a barra correspondente à esquerda.**) 

````python
def print_solution(xs, ys, freq=1):
    """
    Given two arrays xs and ys of the same length, prints a table whose n-th
    line consists of three entries: the values of n, xs[n] and ys[n].
    Parameters:
        * The arrays xs and ys. 
        * A parameter freq used to print only one in every freq line. The
          first, second and last line are always printed. If freq == 0, then
          only these lines are printed.
    Returns: None.
    Prints: A header and the table described above.
    """
    def print_header():
        """
        Prints the table's header.
        """
        print("\n|       n      ", end="")
        print("    x_n            ", end="")
        print("    f(x_n)      |")
        print("|=================================================|")
        
    def print_line(n, x, y):
        """
        Pretty-prints n, x and y.
        """
        if n == 0:
            print(f"|       a", end="")
        elif n == 1:
            print(f"|       b", end="")
        else:
            print(f"|      {n - 1:02}", end="")
        print(f"    {x:15.8f}", end="")
        print(f"    {y:15.8f}   |")
    
    
    N = len(xs)
    if freq == 0:       # If freq == 0, print only first and last lines.
        freq = N - 1
    print_header()
    for n in range(0, 2):
        print_line(n, xs[n], ys[n])
    for n in range(2, N, freq):
        print_line(n, xs[n], ys[n])
    if n != N - 1:
        print_line(n, xs[N - 1], ys[N - 1])
    print("|_________________________________________________|\n")
        
    return None
````

````python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

def bisection_animation(f, a, b, N=4, title="", duration=0.75):
    """
    Displays an animation of the bisection method applied to a function.

    Parameters:
    * f: A continuous real function f.
    * a: The left endpoint of the interval [a, b] such that f(a)f(b) < 0.
    * b: The right endpoint of the interval [a, b] such that f(a)f(b) < 0.
    * N: The maximum number of iterations.
    * title: A title to be displayed at the top of the diagram.
    * duration: The duration of the pause between slides of the animation, in seconds.

    Returns:
    * xs: List of float, the estimates from the bisection method.
    * ys: List of float, the values of the function f at each of the estimates.
    * animation_html: IPython.display.HTML object containing the animation.
    """

    def iterate(a, b):
        """Applies a single step of the bisection method to the interval [a, b]."""
        m = 0.5 * (a + b)  # Midpoint of [a, b]
        if np.sign(f(a)) != np.sign(f(m)):  # [a, m] contains a zero
            return m, a, m
        else:  # [m, b] contains a zero
            return m, m, b

    cmap = plt.get_cmap("tab10")
    P = 200
    width = 1.75
    marker_size = 5
    domain = np.linspace(a, b, P)
    xs = [a, b]
    for _ in range(N):
        m, a, b = iterate(a, b)
        xs.append(m)
    ys = [f(x) for x in xs]

    # Lists containing the x and y coordinates for plotting vertical lines:
    xs_vert = [np.linspace(xs[n], xs[n], P) for n in range(N + 2)]
    ys_vert = [np.linspace(0, ys[n], P) for n in range(N + 2)]

    fig, ax = plt.subplots()
    ax.axhline(y=0.0, color='black', linestyle='-', lw=width)
    ax.plot(domain, f(domain), label="$ y = f(x) $", lw=width)
    ax.set_xlabel("$ x $")
    ax.set_ylabel("$ y $")
    ax.set_title(title)
    ax.grid(linestyle="dotted", alpha=0.5)
    ax.legend()

    lines = []

    for i in range(N + 2):
        color = cmap(i + 1)
        (line,) = ax.plot([], [], color=color, marker="x", mew=width)
        (vert_line,) = ax.plot([], [], linestyle='--', label=f"$ x_{i + 1} $", lw=width)
        (point,) = ax.plot([], [], color='black', marker="o", ms=marker_size)
        lines.append((line, vert_line, point))

    def init():
        for line, vert_line, point in lines:
            line.set_data([], [])
            vert_line.set_data([], [])
            point.set_data([], [])
        return [item for sublist in lines for item in sublist]

    def update(frame):
        if frame == 0:
            return [item for sublist in lines for item in sublist]
        n = frame - 1
        if n < N + 2:
            x = xs[n]
            y = ys[n]
            line, vert_line, point = lines[n]
            line.set_data([x], [0])
            vert_line.set_data(xs_vert[n], ys_vert[n])
            point.set_data([x], [y])
        return [item for sublist in lines for item in sublist]

    ani = FuncAnimation(fig, update, frames=range(N + 3), init_func=init,
                        blit=True, interval=duration * 1000)
    
    plt.close(fig)
    return xs, ys, HTML(ani.to_jshtml())
````

````python
a = 0           # Extremidade esquerda do intervalo inicial, onde f vale -3.
b = 2           # Extremidade direita, onde f vale 3.
N = 5           # Número de iterações desejado.
pausa = 1.0     # Intervalo de tempo entre cada passo da animação, em segundos.
f = lambda x: x**3 - x - 3    # Função à qual o método será aplicado.
# Título a ser exibido no topo do diagrama:
titulo = "Método da bissecção para $ y = x^3 - x - 3, a = 0, b = 2 $."

xs, ys, ani = bisection_animation(f, a, b, N, titulo, pausa)
ani
````

## $ \S 3 $ Implementação do método da bissecção

<div style="border:2px solid #444; padding:15px; border-radius:8px; background:#f7f7f7">

### ⚙️ Algoritmo — Método da Bissecção

Seja $f(x)$ contínua em $[a,b]$ tal que $f(a)f(b) < 0$.

**Entrada**

- Intervalo inicial $[a,b]$
- Precisão $\varepsilon > 0$

**Saída**

- Aproximação $\bar{x}$ de uma raiz de $f(x)=0$

---

**Passos**

1. **Dados iniciais**

   a) intervalo inicial $[a,b]$  
   b) precisão $\varepsilon$

2. Se $(b-a) < \varepsilon$, então escolha para $\bar{x}$ qualquer $x \in [a,b]$. **FIM**

3. $k \leftarrow 1$

4. $M \leftarrow f(a)$

5. $x \leftarrow \dfrac{a+b}{2}$

6. Se $M f(x) > 0$, faça  
   &nbsp;&nbsp;&nbsp;&nbsp;$a \leftarrow x$  
   Vá para o passo 8.

7. $b \leftarrow x$

8. Se $(b-a) < \varepsilon$, então escolha para $\bar{x}$ qualquer $x \in [a,b]$. **FIM**

9. $k \leftarrow k + 1$. Volte ao passo 5.

---
Fonte: [Ruggiero & Lopes, 1996](#ref-ruggiero).
</div>

````python
def bisseccao(f, a, b, epsilon):
    """
    Método da bissecção conforme Algoritmo 1.

    Imprime:
    k, x, f(x), b-a
    """

    if f(a) * f(b) > 0:
        raise ValueError("Intervalo inválido: f(a)f(b) deve ser < 0")

    print(f"{'k':>3} {'x':>14} {'f(x)':>14} {'b-a':>14}")
    print("-" * 50)

    # passo 2
    if (b - a) < epsilon:
        print(f"\nIntervalo [{a:.6f}, {b:.6f}] suficientemente pequeno: {erro:.6e} < {epsilon:.6e}.")
        return (a + b) / 2 # Como (b-a) < epsilon, o intervalo já é suficientemente pequeno, o ponto médio é tomado como aproximação da raiz.        
    # passo 3
    k = 1

    # passo 4
    M = f(a)

    while True:

        # passo 5
        x = (a + b) / 2
        fx = f(x)

        # passo 6 e 7
        if M * fx > 0:
            a = x
        else:
            b = x

        erro = abs(b - a)

        # print da iteração
        print(f"{k:3d} {x:14.6f} {fx:14.6e} {erro:14.6e}")

        # passo 8
        if erro < epsilon:            
            print(f"\nIntervalo [{a:.6f}, {b:.6f}] suficientemente pequeno: {erro:.6e} < {epsilon:.6e}.")
            return (a + b) / 2 # Como (b-a) < epsilon, o intervalo já é suficientemente pequeno, o ponto médio é tomado como aproximação da raiz.            
        # passo 9

        k += 1
````

<div style="background-color: #fff3b0; padding: 12px; border-radius: 6px;">

- ⚠️ É recomendável impor um **limite máximo de iterações como critério de parada**, evitando execuções muito longas ou situações em que a precisão desejada não pode ser atingida por limitações numéricas.

- ⚠️ Em computação numérica, não se deve testar diretamente se $f(x)=0$; em vez disso, verifica-se se $|f(x)| < \varepsilon,$ para uma tolerância adequada (que pode ser diferente da usada para o intervalo).
</div>

## §5 — Análise do desempenho do método da bissecção

A cada iteração do método da bissecção, o intervalo que contém a raiz é dividido pela metade.

Se o intervalo inicial é $[a,b]$ então, após $n$ iterações, o comprimento do intervalo será $$\frac{b-a}{2^n}.$$

Como a aproximação da raiz é o **ponto médio do intervalo**, o erro máximo da aproximação é limitado por

$$
\frac{b-a}{2^n}.
$$

Para garantir que o erro seja menor ou igual a um valor desejado $\varepsilon$, precisamos que

$$
\frac{b-a}{2^n} \le \varepsilon.
$$

Isso implica

$$
n \ge \log_2\!\left(\frac{b-a}{\varepsilon}\right).
$$

Portanto, o número mínimo de iterações necessárias é

$$
\boxed{
n =
\left\lceil
\log_2\!\left(\frac{b-a}{\varepsilon}\right)
\right\rceil
}
$$

### Retomando o exemplo [__Exemplo 2-02.2__](#exemplo-2-02.2)

Vamos retomar o exemplo da função $f(x) = x^3 - x - 3$ no intervalo inicial $[0,2].$

Observe que $f(0) = -3 < 0$ e $f(2) = 3 > 0.$

Como a função é contínua, pelo **Teorema de Bolzano** sabemos que existe pelo menos uma raiz no intervalo $[0,2]$.

Nosso objetivo agora é determinar **quantas iterações do método da bissecção são necessárias** para garantir uma aproximação da raiz com erro máximo $\varepsilon = 10^{-5}.$

Lembre que, no método da bissecção, após $n$ iterações o comprimento do intervalo que contém a raiz é 

$$
\frac{b-a}{2^n}.
$$

Portanto, para garantir que o erro seja menor ou igual a $\varepsilon$, devemos escolher $n$ tal que

$$
\frac{b-a}{2^n} \le \varepsilon.
$$

No nosso caso:

- intervalo inicial: $a=0$, $b=2$
- tolerância desejada: $\varepsilon = 10^{-5}$

Assim, vamos calcular o **número mínimo de iterações necessárias** para atingir essa precisão.

````python
import numpy as np

a = 0
b = 2
eps = 1e-5

n = np.ceil(np.log2((b-a)/eps))

print("Número mínimo de iterações:", int(n))
````

<div style="background-color: #D1FFBD; padding: 12px; border-radius: 6px;">

<b>Observação:</b> O método da bissecção apresenta algumas vantagens importantes. 
Primeiramente, sua convergência é garantida (desde que sejam satisfeitas as hipóteses de continuidade de $f(x)$ em $[a,b]$ e de mudança de sinal entre $a$ e $b$), pois a cada iteração a distância até a raiz é reduzida pela metade, tendendo naturalmente a zero. 

Outra vantagem é a possibilidade de estimar, antecipadamente, o número de iterações necessárias para atingir uma precisão desejada.
