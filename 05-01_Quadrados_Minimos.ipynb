## $ \S 1 $ Método dos Mínimos Quadrados (MMQ) — Caso Discreto

Dada uma tabela de pontos:

$$
\begin{array}{|c|c|c|c|c|}
\hline
x_1 & x_2 & x_3 & \cdots & x_m \\
\hline
f(x_1) & f(x_2) & f(x_3) & \cdots & f(x_m) \\
\hline
\end{array}
$$

A função extrapoladora será dada por:

$$
f(x)\approx\phi(x)
=
a_1g_1(x)+a_2g_2(x)+\cdots+a_ng_n(x).
$$

**A distância entre os dados $f(x_k)$ e a função de aproximação $\phi(x_k)$ deve ser minimizada.**

## $ \S 3 $ Formulação MMQ

Considere os dados $(x_k,f(x_k))$, para $k=1,2,\ldots,m$, e a função aproximadora

$$
\phi(x)=a_1g_1(x)+a_2g_2(x)+\cdots+a_ng_n(x).
$$

O Método dos Mínimos Quadrados determina os coeficientes $a_1,a_2,\ldots,a_n$ minimizando

$$
F(a_1,a_2,\ldots,a_n)
=
\sum_{k=1}^{m}\left[f(x_k)-\phi(x_k)\right]^2.
$$

Para isso, fazemos

$$
\frac{\partial F}{\partial a_1}=0,
\quad
\frac{\partial F}{\partial a_2}=0,
\quad
\ldots,
\quad
\frac{\partial F}{\partial a_n}=0.
$$

<div style="border:2px solid #444; padding:15px; border-radius:8px; background:#f7f7f7">

**Equações Normais**

$$
\left\{
\begin{aligned}
\left[\sum_{k=1}^{m}g_1(x_k)g_1(x_k)\right]a_1
+\cdots+
\left[\sum_{k=1}^{m}g_1(x_k)g_n(x_k)\right]a_n
&=\sum_{k=1}^{m}g_1(x_k)f(x_k),\\[5pt]
\left[\sum_{k=1}^{m}g_2(x_k)g_1(x_k)\right]a_1
+\cdots+
\left[\sum_{k=1}^{m}g_2(x_k)g_n(x_k)\right]a_n
&=\sum_{k=1}^{m}g_2(x_k)f(x_k),\\
&\ \vdots\\
\left[\sum_{k=1}^{m}g_n(x_k)g_1(x_k)\right]a_1
+\cdots+
\left[\sum_{k=1}^{m}g_n(x_k)g_n(x_k)\right]a_n
&=\sum_{k=1}^{m}g_n(x_k)f(x_k).
\end{aligned}
\right.
$$

</div>

## $ \S 4 $ Caso polinomial

Para um polinômio de grau $p$, adotamos

$$
g_1(x)=1,\quad g_2(x)=x,\quad \ldots,\quad g_{p+1}(x)=x^p,
$$

logo

$$
\phi(x)=a_1+a_2x+\cdots+a_{p+1}x^p.
$$

As equações normais do polinômio extrapolador são

$$
\begin{bmatrix}
\sum_{k=1}^{m} 1 & \sum_{k=1}^{m} x_k & \cdots & \sum_{k=1}^{m}x_k^p\\
\sum_{k=1}^{m}x_k & \sum_{k=1}^{m}x_k^2 & \cdots & \sum_{k=1}^{m}x_k^{p+1}\\
\vdots & \vdots & \ddots & \vdots\\
\sum_{k=1}^{m}x_k^p & \sum_{k=1}^{m}x_k^{p+1} & \cdots & \sum_{k=1}^{m}x_k^{2p}
\end{bmatrix}
\begin{bmatrix}
a_1\\a_2\\\vdots\\a_{p+1}
\end{bmatrix}
=
\begin{bmatrix}
\sum_{k=1}^{m}f(x_k)\\
\sum_{k=1}^{m}x_k f(x_k)\\
\vdots\\
\sum_{k=1}^{m}x_k^p f(x_k)
\end{bmatrix}.
$$


## $ \S 5 $ Código para ajuste linear

A rotina fornecida realiza o ajuste por uma reta. Com a notação teórica adotada,

$$
\phi(x)=a_1+a_2x,
\qquad
g_1(x)=1,\quad g_2(x)=x.
$$

Nesse caso, as equações normais são escritas como

$$
\begin{bmatrix}
m & \sum_{k=1}^{m}x_k\\
\sum_{k=1}^{m}x_k & \sum_{k=1}^{m}x_k^2
\end{bmatrix}
\begin{bmatrix}
a_1\\a_2
\end{bmatrix}
=
\begin{bmatrix}
\sum_{k=1}^{m}f(x_k)\\
\sum_{k=1}^{m}x_k f(x_k)
\end{bmatrix}.
$$

<div style="background-color:#fff3b0; padding:12px; border-radius:6px;">

**Nota de implementação:** no código, `X` armazena os valores $x_k$, `Y` armazena os valores $f(x_k)$ e o vetor retornado usa posições Python: `a[0]` representa $a_1$ e `a[1]` representa $a_2$.

</div>

```python
def imprimir_funcao_mmq(coeficientes, nome="phi", variavel="x", precisao=6):
    """
    Imprime a funcao polinomial obtida pelo MMQ.

    A posicao 0 representa a_1, a posicao 1 representa a_2, e assim por diante.
    """
    coeficientes = np.asarray(coeficientes, dtype=float)
    termos = []

    for grau, coef in enumerate(coeficientes):
        modulo = abs(coef)
        if grau == 0:
            termo = f"{modulo:.{precisao}g}"
        elif grau == 1:
            termo = f"{modulo:.{precisao}g}{variavel}"
        else:
            termo = f"{modulo:.{precisao}g}{variavel}^{grau}"

        if not termos:
            termos.append(("-" if coef < 0 else "") + termo)
        else:
            termos.append((" - " if coef < 0 else " + ") + termo)

    expressao = "".join(termos)
    print(f"{nome}({variavel}) = {expressao}")
    return expressao
```

```python
def mmq_linear(X, Y):
    """
    Ajusta os dados por uma reta phi(x) = a_1 + a_2*x.
    """
    
    X = np.asarray(X, dtype=float)
    Y = np.asarray(Y, dtype=float)
    m = len(X)

    # Montagem das equacoes normais para g_1(x)=1 e g_2(x)=x.
    A = np.array([[m, np.sum(X)],
                  [np.sum(X), np.sum(X**2)]], dtype=float)
    b = np.array([np.sum(Y), np.sum(X*Y)], dtype=float)

    # Solucao do sistema linear das equacoes normais.
    a = np.linalg.solve(A, b)
    
    print("Equação da reta ajustada:")
    imprimir_funcao_mmq(a)

    # Grafico da reta ajustada e dos dados tabelados.
    margem = 0.05 * (np.max(X) - np.min(X)) if np.max(X) != np.min(X) else 1.0
    eixox = np.linspace(np.min(X) - margem, np.max(X) + margem, 200)
    phi = a[0] + a[1] * eixox

    plt.figure(figsize=(8, 5))
    plt.plot(eixox, phi, "b", label="Reta ajustada")
    plt.scatter(X, Y, color="red", marker="o", s=65, label="Dados tabelados", zorder=3)
    plt.xlabel(r"$x$")
    plt.ylabel(r"$f(x)$")
    plt.grid(True)
    plt.legend()
    plt.show()

    return a
```
