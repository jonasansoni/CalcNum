# 🚀 Aula 0 — Introdução ao Jupyter Notebook

Nesta disciplina utilizaremos **Python** e **Jupyter Notebook** como principais ferramentas para implementar e testar métodos numéricos.

O Jupyter Notebook permite combinar em um único documento:

- 📝 texto explicativo
- 🧮 fórmulas matemáticas

Isso transforma o notebook em um **caderno de laboratório computacional**.

## 📘 O que é o Jupyter Notebook?

O **Jupyter Notebook** é um ambiente de computação interativo que funciona diretamente no navegador.

Ele permite:

- 💻 escrever e executar **código Python**
- 📝 incluir **texto formatado em Markdown**
- 🧮 utilizar **fórmulas matemáticas em LaTeX**
- 📊 gerar **gráficos e visualizações**
- 📑 documentar todo o processo de resolução de um problema

Cada arquivo de notebook possui a extensão: `.ipynb`


Esse arquivo contém todas as células de código, texto e resultados gerados.

* O que inspirou a utilização desta metodologia foi o curso apresentado pela __Profa. Lorena Barba__ (George Washington University, USA) [Practical Numerical Methods with Python](https://github.com/numerical-mooc/numerical-mooc). Material desenvolvido e adaptado a partir do Practical Numerical Methods with Python project "Copyright (c) Barbagroup".

## 🧩 Estrutura de um Notebook

Um notebook é composto por **células**.

### 💻 Células de código (Code)

Executam **código Python**.

Após executar a célula, o resultado aparece logo abaixo.

### 📝 Células de texto (Markdown)

Permitem escrever:

- texto formatado
- listas
- imagens
- fórmulas matemáticas

Elas são usadas para **explicar o que o código está fazendo**.

Esse arquivo contém todas as células de código, texto e resultados gerados durante a execução.

## ▶️ Executando células

Para executar uma célula no Jupyter Notebook:

**Shift + Enter**

Outros comandos úteis:

- **Ctrl + Enter** → executa a célula atual
- **Alt + Enter** → executa e cria uma nova célula

## Executando células

Para executar uma célula no Jupyter Notebook:

- pressione **Shift + Enter**

Esse comando executa a célula atual e seleciona a próxima.

Outros comandos úteis:

- **Ctrl + Enter** → executa a célula sem mudar de posição
- **Alt + Enter** → executa a célula e cria uma nova célula abaixo

## 🖥️ Interface do Jupyter Notebook

A interface possui alguns elementos importantes:

### 📑 Barra de menu
Permite acessar funções como:

- salvar o notebook
- reiniciar o kernel
- executar todas as células

### 🛠️ Barra de ferramentas
Contém botões para:

- executar células
- parar execução
- adicionar células

### 📄 Área do notebook
Região onde aparecem:

- células de código
- células de texto
- resultados das execuções

## 🌐 Onde utilizar o Jupyter Notebook

Você poderá utilizar o Jupyter de duas formas:

## ⚙️ Instalando o Jupyter Notebook

Caso prefira utilizar o Jupyter Notebook diretamente no seu computador, é possível instalar a ferramenta localmente.

As instruções oficiais de instalação estão disponíveis em:

👉 https://jupyter.org/install

Nesse site você encontrará diferentes opções de instalação, incluindo:

- instalação usando **pip**
- instalação usando **conda**
- instalação do **JupyterLab** ou **Jupyter Notebook**

Após a instalação, você pode iniciar o Jupyter executando no terminal:
`jupyter notebook`



O ambiente será aberto automaticamente no navegador.


## Teste sua instalação

### Inciando o _notebook server_

Abra o terminal e execute

```Bash
jupyter notebook
```
Isto irá iniciar o notebook server. Isto * deve * abrir automaticamente a página principal do servidor do notebook no seu navegador padrão, mas se isso não acontecer, basta abrir o navegador de sua escolha e digitar

```
http://localhost:8888/tree
```

na barra de endereço.

Isso trará um navegador de arquivos simples que mostrará o conteúdo do diretório de onde você iniciou o terminal.  Click em `New Notebook` e então selecione **Python 3** para criar o seu primeiro jupyter notebook.

## ☁️ Executando Jupyter online

Também é possível utilizar o Jupyter Notebook **sem instalar nenhum programa**, diretamente no navegador.  
Abaixo estão algumas opções gratuitas:

### 🚀 Try Jupyter (Jupyter.org)

Experimente o Jupyter diretamente no navegador usando o ambiente oficial do projeto Jupyter:

Executa notebooks diretamente de um repositório online.

O Binder cria automaticamente um ambiente completo para executar notebooks diretamente no navegador.

---

### 🔗 Google Colab

Ambiente Jupyter mantido pelo Google com suporte a Python e bibliotecas científicas.

👉 https://colab.research.google.com

---

⚠️ **Importante**

Ambientes online geralmente são **temporários**.  
Se você modificar um notebook, recomenda-se **baixar o arquivo para o seu computador** antes de encerrar a sessão.

### Gráfico simples

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,10,100)
y = np.sin(x)

plt.plot(x,y)
plt.title("Exemplo de gráfico")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
````
